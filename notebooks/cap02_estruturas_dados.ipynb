## [1.1] Categorias de dados

**Dados** são medidas, contagens, observações, características ou atributos de um estado ou objeto, que por sí só não carregam significado nem informação.

Os dados se dividem em três principais categorias:

- [[dados não estruturados]]
- [[dados estruturados]]
- [[dados semiestruturados]]

Os dados também pode ser de [[séries temporais]] um tipo especial, que podem ser **estruturados** ou **semiestruturados**.

Os **dados** se originam de diversas [[fontes de dados]] diferentes, incluíndo textos, vídeos, imagens, sensores de dispositivos, entre outros.
