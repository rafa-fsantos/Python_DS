## 2. Estrutura de Dados: Listas

Listas em Python são sequências mutáveis, ordenadas e indexadas.


```python
# Criação de lista com elementos iniciais
regions = ['Asia', 'America', 'Europe']
print("Lista de regiões:", regions)

# Criação de uma lista vazia para gerenciamento de tarefas
my_list = []

# Adicionando elementos ao final da lista sequencialmente
my_list.append('Pay bills')
my_list.append('Tidy up')
my_list.append('Walk the dog')
my_list.append('Cook dinner')

print("Lista inicial de tarefas:", my_list)
```

```python
# Acesso direto a elementos por índice posicional (índice 0 = primeiro elemento)
print("Primeira tarefa:", my_list[0])

# Localizando a posição (índice) de um valor específico
i = my_list.index('Cook dinner')
print(f"Índice de 'Cook dinner': {i}")

# Inserindo um item em uma posição arbitrária antes de 'Cook dinner'
my_list.insert(i, 'Go to pharmacy')
print("Após inserção:", my_list)

# Contagem de ocorrências de um item na lista
qtd_tidy_up = my_list.count('Tidy up')
print(f"Ocorrências de 'Tidy up': {qtd_tidy_up}")
```

### 2.1 Notação de Fatiamento (Slicing)

Sintaxe: `lista[início:fim:passo]` (o limite superior `fim` é exclusivo).

```python
# Fatiamento dos três primeiros elementos (índices 0, 1 e 2)
print("Primeiros três itens [0:3]:", my_list[0:3])

# Omitindo o início (assume-se 0 implicitamente)
print("Primeiros três itens [:3]:", my_list[:3])

# Omitindo o fim (continua até o último item a partir do índice 3)
print("A partir do índice 3 [3:]:", my_list[3:])

# Cópia superficial (shallow copy) da lista inteira
print("Cópia da lista [:]:", my_list[:])
```

```python
# Atribuição por fatia: adicionando múltiplos elementos ao final
my_list[len(my_list):] = ['Mow the lawn', 'Water plants']
print("Lista expandida por slice:", my_list)

# Remoção de itens por fatia utilizando o operador 'del'
del my_list[5:]
print("Lista após remoção via del:", my_list)
```

## 3. Estruturas Abstratas de Dados

### 3.1 Filas (Queues - FIFO: First-In, First-Out)

Na fila, o primeiro item a entrar é o primeiro a sair. Embora uma lista padrão suporte `.pop(0)`, essa operação tem complexidade $O(n)$. O objeto `collections.deque` oferece inserções e remoções eficientes nas duas pontas em $O(1)$.


```python
# Inicializando uma fila de tarefas com deque
queue = deque(my_list)

# Inserção no final da fila (enqueue)
queue.append('Wash the car')

# Remoção da cabeça da fila (dequeue)
tarefa_concluida = queue.popleft()
print(f"Concluída: '{tarefa_concluida}'")

# Convertendo de volta para lista caso necessário
my_list_upd = list(queue)
print("Estado atualizado da fila:", my_list_upd)
```

### 3.2 Pilhas (Stacks - LIFO: Last-In, First-Out)

Em uma pilha, o último item adicionado é o primeiro a ser consumido. As listas nativas do Python funcionam como pilhas eficientes através dos métodos `.append()` e `.pop()`.


```python
tarefas_pilha = ['Pay bills', 'Tidy up', 'Walk the dog', 'Go to the pharmacy', 'Cook dinner']
stack = []

# Empilhando (Push)
for task in tarefas_pilha:
    stack.append(task)

print("Iniciando o desempilhamento:")
# Desempilhando (Pop) até esvaziar
while stack:
    item = stack.pop()
    print(f"- {item} (Processado)")

print("A pilha está vazia.")
```

## 4. Aplicação: Processamento de Linguagem Natural (PLN)

Uso de listas e pilhas para agrupar sintagmas nominais (*noun phrases*) através da árvore de dependência do spaCy.

```python
txt = 'List is a ubiquitous data structure in the Python programming language.'
doc = nlp(txt)
stk = []

print("Extração de termos com pilha básica:")
for w in doc:
    # Se o token for um substantivo ou nome próprio
    if w.pos_ in ('NOUN', 'PROPN'):
        stk.append(w.text)
    # Se for um modificador que aponta para a esquerda de um substantivo
    elif (w.head.pos_ in ('NOUN', 'PROPN')) and (w in w.head.lefts):
        stk.append(w.text)
    # Quando o sintagma termina, desempilha para formar a expressão
    elif stk:
        chunk = ''
        while stk:
            chunk = stk.pop() + ' ' + chunk
        print("  ->", chunk.strip())

# Esvazia qualquer elemento remanescente na pilha após o loop
if stk:
    chunk = ''
    while stk:
        chunk = stk.pop() + ' ' + chunk
    print("  ->", chunk.strip())
```

### 4.1 Explorando Relações Sintáticas com List Comprehensions

```python
txt_exemplo = 'List is arguably the most useful type in the Python programming language.'
doc_exemplo = nlp(txt_exemplo)

# Visualizando pares (token, seu núcleo sintático)
print("Tokens e seus respectivos núcleos:")
for t in doc_exemplo:
    print(f"  {t.text:<12} -> {t.head.text}")
```

```python
# Marcando tokens dependentes à esquerda (lefts)
head_lefts = [t.text if t in t.head.lefts else 0 for t in doc_exemplo]
print("\nTokens dependentes à esquerda:")
print(head_lefts)
```

### 4.2 Modelo Final de Extração de Sintagmas com Suporte de Pilha

```python
doc_final = nlp(txt_exemplo)
stk = []

print("Sintagmas extraídos pelo modelo consolidado:")
for w in doc_final:
    # Vetor binário de dependentes à esquerda a partir da posição atual
    head_lefts = [1 if t in t.head.lefts else 0 for t in doc_final[w.i:]]
    
    i0 = 0
    try:
        i0 = head_lefts.index(0)
    except ValueError:
        pass

    i1 = 0
    if i0 > 0:
        # Busca reversa pelo substantivo na janela selecionada
        noun_window = [1 if t.pos_ in ('NOUN', 'PROPN') else 0 for t in reversed(doc_final[w.i:w.i + i0 + 1])]
        try:
            i1 = noun_window.index(1) + 1
        except ValueError:
            pass

    # Critérios de empilhamento
    if w.pos_ in ('NOUN', 'PROPN') or i1 > 0:
        stk.append(w.text)
    elif stk:
        chunk = ''
        while stk:
            chunk = stk.pop() + ' ' + chunk
        print("  ->", chunk.strip())

# Limpeza e exibição do último sintagma retido
if stk:
    chunk = ''
    while stk:
        chunk = stk.pop() + ' ' + chunk
    print("  ->", chunk.strip())
```
