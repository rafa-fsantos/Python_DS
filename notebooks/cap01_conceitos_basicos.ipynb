## 📘 Capítulo 1: Conceitos Básicos

**Dados** são medidas, contagens, observações, características ou atributos de um estado ou objeto. Isoladamente, não carregam significado nem informação contextual. Eles se originam de diversas fontes — como textos, vídeos, imagens e sensores de dispositivos — e dividem-se em três categorias centrais, além de uma categoria temporal específica.

---

### 🏷️ Categorias de Dados

#### 🧩 Dados Não Estruturados

* **Definição:** Informações que não possuem estrutura organizacional ou esquema predefinido.
* **Prevalência:** Representam o tipo mais comum de dado existente.
* **Exemplos:** Imagens, vídeos, áudios e textos em linguagem natural.
* **Aproveitamento:** Podem ser processados e convertidos em dados estruturados ou semiestruturados.

#### 📊 Dados Estruturados

* **Definição:** Dados organizados sob um formato rígido e padronizado.
* **Armazenamento:** Comumente guardados em bancos de dados relacionais ou em arquivos tabulares (como `.csv` e `.xlsx`).
* **Composição:** Atributos e características ocupam **campos** (colunas), enquanto os registros correspondentes formam as **linhas**.
* **Tipos principais:**
  * 🔤 **Categóricos:** Agrupam itens com base em características compartilhadas. Podem assumir representações numéricas (como telefones ou CEPs), mas não comportam operações matemáticas ou estatísticas (como média ou mediana).
  * 🔢 **Numéricos:** Expressam grandezas quantitativas e suportam cálculos matemáticos diretamente.

#### 🗂️ Dados Semiestruturados

* **Definição:** Dados que não seguem tabelas rígidas, mas contêm marcadores organizacionais internos ou tags autodescritivas.
* **Formatos usuais:** Arquivos nos formatos `JSON` e `XML`.

#### 📈 Dados de Séries Temporais

* **Definição:** Conjuntos de observações ordenadas e indexadas cronologicamente ao longo de um período específico.
* **Classificação:** Podem se apresentar tanto em formatos **estruturados** quanto **semiestruturados**.

---

### 🌐 Fontes de Dados

No contexto de Data Science e desenvolvimento em Python, as principais fontes para obtenção de dados são:

#### 🔌 APIs (Interfaces de Programação de Aplicações)

* **Papel:** Softwares intermediários que permitem a comunicação direta entre sistemas distintos. Representam uma das fontes mais populares e abrangentes.
* **Uso em Python:** Acesso via bibliotecas dedicadas (*wrappers*) ou requisições HTTP por meio do pacote `Requests`.
* **Critérios de escolha:**
  * ⚙️ **Funcionalidade:** Verificar se a ferramenta cobre os requisitos necessários do projeto.
  * 💰 **Custo:** Identificar limitações em opções gratuitas ou custos de planos pagos.
  * 🛡️ **Estabilidade:** Avaliar a confiabilidade e o suporte da biblioteca consultando o repositório PyPI.
  * 📖 **Documentação:** Priorizar soluções bem documentadas, com exemplos claros de comandos e integração.
* **Formatos de saída comuns:** As APIs costumam responder em `JSON`, `XML` ou `CSV`, facilitando a posterior conversão para estruturas analíticas como o `DataFrame` do Pandas.

#### 🕸️ Páginas Web

* **Características:** Podem ser estáticas ou geradas dinamicamente a partir da interação do usuário, reunindo conteúdos heterogêneos.
* **Extração:** Coletadas por meio de técnicas de *web scraping*, sendo fontes valiosas quando bem estruturadas.

#### 🗄️ Bancos de Dados

* **Finalidade:** Ambientes focados em armazenamento, consulta e manipulação eficiente de grandes volumes de informação.
* **Modelos:**
  * 🔗 **Relacionais:** Estruturas convencionais e rígidas que utilizam a linguagem SQL para consultas e gerenciamento.
  * 🔄 **NoSQL:** Modelos flexíveis (*Not Only SQL*) voltados a escalabilidade e variedade de formatos, baseando-se em esquemas como chave-valor, documentos ou grafos.

#### 📁 Arquivos

* **Uso:** Meio amplamente utilizado para persistência local ou compartilhamento de dados nos formatos estruturados, semiestruturados ou não estruturados.

---

### 🔄 Pipeline de Processamento de Dados

O pipeline compreende o ciclo completo de obtenção, preparação e extração de valor a partir dos dados. É estruturado em cinco etapas consecutivas:

* 📥 **1. Aquisição:** Coleta inicial dos dados brutos a partir das fontes selecionadas.
* 🧹 **2. Limpeza:** Detecção, filtragem e correção de registros corrompidos, ausentes ou inconsistentes, garantindo a integridade da base.
* ⚙️ **3. Transformação:** Conversão de formatos, normalização e enriquecimento da estrutura dos dados para viabilizar a análise.
* 💡 **4. Análise:** Etapa central do processo, responsável por interpretar as informações tratadas e extrair conclusões práticas.
* 💾 **5. Armazenamento:** Salvamento dos resultados analíticos em arquivos ou bancos de dados para consumo posterior.

### 🐍 Abordagem Pythônica

Na Ciência de Dados, programar de forma "pythônica" significa construir códigos concisos, legíveis e eficientes. Uma das ferramentas mais expressivas dessa filosofia são as *list comprehensions*, que condensam lógicas complexas de processamento em uma única linha (*one-liner*).

#### Exemplo Prático: Limpeza e Tokenização de Texto

Considere o seguinte trecho textual:

```python
txt = '''Eight dollars a week or a million a year - what is the difference? A mathematician or
a wit would give you the wrong answer. The magi brought valuable gifts, but that was not among
them. - The Gift of the Magi, O'Henry'''
```

Para separar o parágrafo em frases, isolar cada palavra em listas individuais e eliminar pontuações indesejadas, aplica-se uma *list comprehension* aninhada:

```python
word_lists = [[w.replace(',', '') for w in line.split() if w not in ['-']]
              for line in txt.replace('?', '.').split('.')]

print(word_lists)
```
